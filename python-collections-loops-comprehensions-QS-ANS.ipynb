**Question 22**

**Difficulty:** Hard

**Problem Statement:** Given a list of student scores, find and print the second highest score without using `max()` twice on the full list.

**Requirements:**
- Create `scores = [88, 95, 72, 95, 85, 91, 72, 88]`.
- Use `sort()` and slicing or indexing to find the second distinct highest score.
- Print the second highest score.

**Example:**  
`Input: Input: scores = [88, 95, 72, 95, 85, 91, 72, 88]`  
`Output: Output: 91`


````python
scores = [88, 95, 72, 95, 85, 91, 72, 88]

# Sort descending so the highest values come first
scores.sort(reverse=True)

# Index 0 is the highest, index 1 could repeat it, so pick the
# first value that's strictly smaller than the max to be safe
highest = scores[0]
second_highest = next(s for s in scores if s < highest)

print(second_highest)
````

**Difficulty:** Hard

**Problem Statement:** Given a list of words, group them by their first letter using a nested list approach.

**Requirements:**
- Create `words = ['apple','avocado','banana','blueberry','cherry','apricot','coconut']`.
- Create three separate lists: `a_words`, `b_words`, `c_words`.
- Traverse `words` and use indexing `[0]` to check the first letter.
- Use `append()` to place each word in the correct group.
- Print all three grouped lists.

**Example:**  
`Input: Input: words = ['apple','avocado','banana','blueberry','cherry','apricot','coconut']`  
`Output: Output: a_words: ['apple','avocado','apricot']  b_words: [...]  c_words: [...]`


````python
words = ['apple', 'avocado', 'banana', 'blueberry', 'cherry', 'apricot', 'coconut']

a_words = []
b_words = []
c_words = []

# Group each word by its first letter
for word in words:
    if word[0] == 'a':
        a_words.append(word)
    elif word[0] == 'b':
        b_words.append(word)
    else:
        c_words.append(word)

print("a_words:", a_words)
print("b_words:", b_words)
print("c_words:", c_words)
````

**Question 25**

**Difficulty:** Hard

**Problem Statement:** Given two lists, find the common elements (intersection) and the elements present in only one list (symmetric difference) without using sets.

**Requirements:**
- Create `list_a = [1, 2, 3, 4, 5, 6]` and `list_b = [4, 5, 6, 7, 8, 9]`.
- Traverse `list_a` and check membership in `list_b` using `in` to find common elements.
- Collect elements that are only in one of the lists.
- Print `common` and `unique_to_one`.

**Example:**  
`Input: Input: list_a=[1,2,3,4,5,6], list_b=[4,5,6,7,8,9]`  
`Output: Output: common: [4,5,6]   unique_to_one: [1,2,3,7,8,9]`


````python
list_a = [1, 2, 3, 4, 5, 6]
list_b = [4, 5, 6, 7, 8, 9]

common = []
unique = []

# Elements of list_a that are (or aren't) also in list_b
for i in list_a:
    if i in list_b:
        common.append(i)
    else:
        unique.append(i)

# Elements of list_b that are not in list_a
for i in list_b:
    if i not in list_a:
        unique.append(i)

print("common:", common)
print("unique:", unique)
````

**Problem Statement:** Perform a multi-step set analysis: given survey data from 3 groups, find who participated in exactly 2 surveys.

**Requirements:**
- Create `survey1 = {'A','B','C','D','E'}`, `survey2 = {'B','C','F','G','A'}`, `survey3 = {'C','D','G','H','B'}`.
- Find participants in all 3 surveys.
- Find participants in exactly 2 surveys (in any two but not all three).
- *(Hint: Use union, intersection, and difference combinations.)*
- Print both results.

**Example:**  
`Input: Input: survey1, survey2, survey3`  
`Output: Output: in_all: {'B','C'}   exactly_two: {'A','D','G'}`


````python
survey1 = {'A', 'B', 'C', 'D', 'E'}
survey2 = {'B', 'C', 'F', 'G', 'A'}
survey3 = {'C', 'D', 'G', 'H', 'B'}

# Participants who took part in all three surveys
in_all = survey1.intersection(survey2).intersection(survey3)

# Combine every pairwise intersection to get "at least 2 surveys"
pair12 = survey1.intersection(survey2)
pair13 = survey1.intersection(survey3)
pair23 = survey2.intersection(survey3)
at_least_two = pair12.union(pair13).union(pair23)

# "Exactly 2" = at least two, minus the ones who are also in all three
exactly_two = at_least_two.difference(in_all)

print("In all three surveys:", in_all)
print("Exactly two surveys:", exactly_two)
````

**Question 73**

**Difficulty:** Medium

**Problem Statement:** Build a word frequency counter for a sentence using a dictionary.

**Requirements:**
- Create `sentence = 'to be or not to be that is the question to ask'`.
- Split into words using `split()`.
- Create `word_count = {}`.
- Count frequency of each word.
- Print the dictionary and find the most frequent word by iterating over items.

**Example:**  
`Input: Input: sentence string`  
`Output: Output: word_count dict   Most frequent: 'to' (count: 3)`


````python
sentence = 'to be or not to be that is the question to ask'

words = sentence.split()
print(words)

word_count = {}

# Build a frequency dictionary word by word
for word in words:
    if word in word_count:
        word_count[word] += 1
    else:
        word_count[word] = 1

print()
print(word_count)

# Find the most frequent word by scanning the dictionary
most_frequent_word = ""
highest_count = 0

for word, count in word_count.items():
    if count > highest_count:
        highest_count = count
        most_frequent_word = word

print("Most frequent word:", most_frequent_word, "-> appears", highest_count, "times")
````

**Question 77**

**Difficulty:** Hard

**Problem Statement:** Group a list of transactions by category using a dictionary of lists.

**Requirements:**
- Create `transactions = [('food',250),('travel',1500),('food',180),('shopping',3200),('travel',800),('food',450),('shopping',900)]`.
- Create `grouped = {}`.
- Traverse `transactions`: if category in `grouped`, append amount; else create a new list.
- Print each category with its list of amounts and total.
- Print the category with the highest total.

**Example:**  
`Input: Input: transactions list`  
`Output: Output: food: [250,180,450] total=880 ...  Highest: shopping`


````python
transactions = [
    ('food', 250),
    ('travel', 1500),
    ('food', 180),
    ('shopping', 3200),
    ('travel', 800),
    ('food', 450),
    ('shopping', 900)
]

grouped = {}

# Group amounts under their category
for category, amount in transactions:
    if category in grouped:
        grouped[category].append(amount)
    else:
        grouped[category] = [amount]

print(grouped)
print()

# Print each category with its amounts and running total
for category, amounts in grouped.items():
    print(f"{category}: {amounts} --> total amount: {sum(amounts)}")

# Find the category with the highest total spending
top_category = ""
top_total = 0

for category, amounts in grouped.items():
    if sum(amounts) > top_total:
        top_total = sum(amounts)
        top_category = category

print(f"\nHighest spending category: {top_category} (Total: {top_total})")
````

**Question 79**

**Difficulty:** Hard

**Problem Statement:** Implement a simple grade book: store student-subject-marks as nested dictionaries and compute each student's average.

**Requirements:**
- Create `gradebook = {'Riya':{'Math':85,'Science':92,'English':88},'Dev':{'Math':70,'Science':78,'English':82},'Sia':{'Math':95,'Science':90,'English':97}}`.
- Iterate using `items()` for outer and inner loops.
- Compute each student's average from their marks.
- Print `Student: Riya | Avg: 88.33` format for each student.
- Find and print the top student.

**Example:**  
`Input: Input: gradebook nested dict`  
`Output: Output: each student's avg   Top: Sia (94.0)`


````python
gradebook = {'Riya': {'Math': 85, 'Science': 92, 'English': 88},
             'Dev': {'Math': 70, 'Science': 78, 'English': 82},
             'Sia': {'Math': 95, 'Science': 90, 'English': 97}}

top_student = ""
highest_avg = 0

# Outer loop: each student. Inner loop: each subject's marks.
for student, subjects in gradebook.items():

    total = 0
    count = 0

    for subject, marks in subjects.items():
        total += marks
        count += 1

    average = total / count

    print(f"Student: {student} | Avg: {round(average, 2)}")

    # Track the student with the highest average so far
    if average > highest_avg:
        highest_avg = average
        top_student = student

print(f"\nTop student: {top_student} with average {round(highest_avg, 2)}")
````

**Question 80**

**Difficulty:** Hard

**Problem Statement:** Merge two dictionaries: for keys present in both, add their values; for unique keys, keep as-is.

**Requirements:**
- Create `store_a = {'apples':30,'bananas':50,'mangoes':20,'grapes':40}`.
- Create `store_b = {'bananas':25,'mangoes':35,'kiwi':60,'oranges':45}`.
- Create `merged = {}`.
- For each key in `store_a`, add to merged.
- For each key in `store_b`, if key exists in merged, add values; else insert.
- Print `merged`.

**Example:**  
`Input: Input: store_a, store_b`  
`Output: Output: {'apples':30,'bananas':75,'mangoes':55,'grapes':40,'kiwi':60,'oranges':45}`


````python
store_a = {'apples': 30, 'bananas': 50, 'mangoes': 20, 'grapes': 40}
store_b = {'bananas': 25, 'mangoes': 35, 'kiwi': 60, 'oranges': 45}

merged = {}

# Copy every item from store_a first
for fruit, quantity in store_a.items():
    merged[fruit] = quantity

# For store_b: add to an existing key, otherwise insert as new
for fruit, quantity in store_b.items():
    if fruit in merged:
        merged[fruit] += quantity
    else:
        merged[fruit] = quantity

print(merged)
````

**Example:**

**Requirements:**
- Create `arr = [64, 34, 25, 12, 22, 11, 90]`.

- Inner loop compares adjacent elements and swaps if out of order.

- Print the sorted list.

**Example:**  
`Input: Input: arr = [64,34,25,12,22,11,90]`  
`Output: Output: [11, 12, 22, 25, 34, 64, 90]`


````python
arr = [64, 34, 25, 12, 22, 11, 90]

swapped = True

# Keep making passes through the list until a pass makes no swaps
while swapped:
    swapped = False
    i = 0

    while i < len(arr) - 1:
        if arr[i] > arr[i + 1]:
            arr[i], arr[i + 1] = arr[i + 1], arr[i]  # swap
            swapped = True
        i += 1

print(arr)
````
