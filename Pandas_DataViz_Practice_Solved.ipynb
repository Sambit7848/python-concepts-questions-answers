### Q67.
Create a copy of the Titanic dataset.

### Q68.
Using a fresh copy of the Titanic dataset, apply **forward fill** using `ffill()` to fill all null values.
Print `isnull().sum()` before and after to compare.


````python
titanic_ffill = titanic.copy()

print("Before ffill:")
print(titanic_ffill.isnull().sum())

# Forward fill: replaces NaN with the previous row's value
titanic_ffill = titanic_ffill.ffill()

print("\nAfter ffill:")
print(titanic_ffill.isnull().sum())
````

### Q69.

Print `isnull().sum()` before and after to compare.


````python
titanic_bfill = titanic.copy()

print("Before bfill:")
print(titanic_bfill.isnull().sum())

# Backward fill: replaces NaN with the next row's value
titanic_bfill = titanic_bfill.bfill()

print("\nAfter bfill:")
print(titanic_bfill.isnull().sum())
````

**Final Challenge:** Build a complete mini analysis on the Titanic dataset:
1. Print shape and `describe()` of the dataset
2. Check null values and fill the `age` column using `ffill()`
3. Use `loc` to filter **female passengers who survived**
4. Use `iloc` to pick the first 10 rows from the filtered result
5. Create a **2×2 subplot** with:
   - Histogram of `age`
   - Bar chart of `pclass` counts
   - Box plot of `fare`
   - Scatter plot of `age` vs `fare`
