## Question 29

Create a 3-D array of shape `(3, 2, 3)` and flatten it to a 1-D array using a loop (without using `.flatten()` or `.reshape()`). Print the original and the resulting 1-D array.

**Hint:**  
Use a triple nested loop and `append()` to collect elements.


````python
# Create a 3-D array of shape (3, 2, 3).
arr = np.arange(1, 19).reshape(3, 2, 3)
print(arr)

print()

# Flatten it to a 1-D array using a triple nested loop.
flat = []
for block in arr:
    for row in block:
        for value in row:
            flat.append(value)

flat = np.array(flat)
print(flat)
````

## Question 43

Create a 1-D array using `np.arange(1, 25)`. Reshape it manually (do not use `.reshape()`) into a 3-D structure by slicing it into a nested list and converting back with `np.array()`. The final shape should be `(2, 3, 4)`.

**Hint:**  
Slice the flat array into segments and nest them into lists of the right dimensions.


````python
# Create a 1-D array.
flat = np.arange(1, 25)
print(flat)

print()

# Manually reshape into a 3-D structure of shape (2, 3, 4).
nested = []
for i in range(2):
    block = []
    for j in range(3):
        start = (i * 12) + (j * 4)
        block.append(list(flat[start:start + 4]))
    nested.append(block)

arr_3d = np.array(nested)
print(arr_3d)
print(arr_3d.shape)
````

Create a 6×6 matrix using `np.arange(1, 37).reshape(6, 6)` — but since you have not used `.reshape()`, do it manually by slicing `np.arange(1, 37)` into 6 segments of 6 and stacking them. Then:
- Print only the border elements (first row, last row, first column, last column)
- Print only the inner 4×4 sub-matrix

**Hint:**  
Slice rows and columns independently to isolate the border and inner sections.


````python
# Manually build a 6x6 matrix by slicing arange(1, 37) into 6 segments.
flat = np.arange(1, 37)
rows = [flat[i * 6:(i + 1) * 6] for i in range(6)]
matrix = np.array(rows)
print(matrix)

print()

# Print only the border elements.
print("First row:", matrix[0, :])
print("Last row:", matrix[-1, :])
print("First column:", matrix[:, 0])
print("Last column:", matrix[:, -1])

print()

# Print only the inner 4x4 sub-matrix.
print(matrix[1:5, 1:5])
````

## Question 49

Using `np.random.shuffle()`, demonstrate that shuffle modifies the array in place:
1. Create an array `a = np.arange(1, 11)`
2. Assign `b = a` (not a copy)
3. Shuffle `a`
4. Print both `a` and `b` and explain why both changed

Then repeat using a proper copy (using `list()` or slicing) so that the original is preserved.

**Hint:**  
Assignment in NumPy creates a reference, not a copy. Use `a.copy()` or `list(a)` for a true copy.


````python
# Create an array.
a = np.arange(1, 11)

# Assign b = a (not a copy).
b = a

# Shuffle a.
np.random.shuffle(a)

# Both a and b changed because b references the same array as a.
print("a:", a)
print("b:", b)

print()

# Repeat using a proper copy.
c = np.arange(1, 11)
d = c.copy()

np.random.shuffle(c)

# Only c changed because d is an independent copy.
print("c:", c)
print("d:", d)
````

## Question 50

Build a mini data simulation pipeline using only the NumPy concepts from this course:
1. Use `np.random.normal(70, 10, (5, 6))` to generate a 5×6 marks matrix (5 students, 6 subjects)
2. Subtract 70 from every element (center around 0)
3. Divide every element by 10 (scale)
4. Use a nested loop to print each student's row with their original subject marks
5. Also print the shape, size, ndim, and dtype of the original matrix

**Hint:**  
Combine random generation, scalar operations, array attributes, and looping over 2-D arrays.


````python
# Generate a 5x6 marks matrix (5 students, 6 subjects).
marks = np.random.normal(70, 10, (5, 6))

# Subtract 70 from every element (center around 0).
centered = marks - 70

# Divide every element by 10 (scale).
scaled = centered / 10

print(scaled)

print()

# Print each student's row with their original subject marks.
for i, row in enumerate(marks):
    print("Student", i + 1, ":", row)

print()

# Print the shape, size, ndim, and dtype of the original matrix.
print(marks.shape)
print(marks.size)
print(marks.ndim)
print(marks.dtype)
````
